```python
# 21, sorted Dictionary by values
dic = {'apple': 50, 'banana': 10, 'cherry': 30}
items = list(dic.items())

for i in range(len(dic)):
    for j in range(0,len(dic)-i - 1):

        if items [j][1]> items[j+1][1]:
            temp = items[j]
            items[j]= items[j+1]
            items[j+1] =temp

sorted_dic = {}

for key, value in items:
    sorted_dic[key] = value
print(sorted_dic)
```
